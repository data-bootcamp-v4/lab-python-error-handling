````python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {products}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory
````

````python
def calculate_total_price(lista_pedidos):
    
    precios = []
    products_dict = {"product_1": "hat", "product_2": "book", "product_3": "mug"}

    for clave in lista_pedidos:
        nombre_producto = products_dict.get(clave, "Producto desconocido")

        while True:
            try:
                precio = float(input(f"Ingrese el precio del producto '{nombre_producto}': "))
                if precio < 0:
                    print("❌ El precio no puede ser negativo. Intente nuevamente.")
                    continue
                precios.append(precio)
                break
            except ValueError:
                print("Entrada inválida. Ingrese un número válido.")
    
    total_price = sum(precios)
    return total_price

````

````python
def get_customer_orders(inventario_inicial):
    while True:
        try:
            num_orders = int(input("¿Cuántos productos desea agregar? "))
            if num_orders <= 0:
                print("❌ El número debe ser mayor que cero.")
                continue
            break
        except ValueError:
            print("Entrada inválida. Debe ingresar un número entero.")

    orders = []

    for i in range(num_orders):
        while True:
            product = input(f"Ingrese el nombre del producto #{i + 1}: ").strip().lower()
            
            if product not in inventario_inicial:
                print("Ese producto no está en el inventario.")
            elif inventario_inicial[product] <= 0:
                print("No hay stock disponible para ese producto.")
            else:
                orders.append(product)
                break  

    return orders
````

````python
products = ["t-shirt", "mug", "book", "hat", "keychain"]
inventario_inicial = initialize_inventory(products)
print("Inventario inicial", inventario_inicial)

lista_pedidos = get_customer_orders(inventario_inicial)
print("Pedidos ingresados:", lista_pedidos)

lista_pedidos = ["product_1", "product_2", "product_3"]
precio_total = calculate_total_price(lista_pedidos)
print("Precio total:", precio_total)
````
