```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def initialise_inventory(products):
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity >= 0:
                    inventory[product] = quantity
                    valid_input = True
                else:
                    print("Quantity cannot be negative. Please enter a valid quantity.")
            except ValueError:
                print("Invalid input. Please enter a valid quantity.")
    return inventory

    customer_orders = set()

    more_order = "yes"

def get_customer_orders(inventory):
    while True:
        try:
            number_orders = int(input("Enter the number of customer orders: "))
            if number_orders < 0:
                print("Number of orders cannot be negative. Please enter a valid number.")
            else:
                break
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

    customer_orders = set()

    for i in range(number_orders):
        while True:
            order = input(f"Enter product name for order {i+1}: ").strip().lower()
            if order not in inventory:
                print("Product not found in inventory. Please enter a valid product.")
            elif inventory[order] <= 0:
                print(f"Sorry, {order} is out of stock. Please choose another product.")
            else:
                customer_orders.add(order)
                break

    return customer_orders

def calculate_total_price(customer_orders):
    prices = {}
    for product in customer_orders:
        valid_input = False
        while not valid_input:
            try:
                price = float(input(f"Enter the price for {product}: "))
                if price < 0:
                    print("Price cannot be negative. Please enter a valid price.")
                else:
                    prices[product] = price
                    valid_input = True
            except ValueError:
                print("Invalid input. Please enter a numeric value.")
    total_price = sum(prices[product] for product in customer_orders)
    return total_price

def update_inventory(customer_orders, inventory):
    updated_inventory = {
        item: inventory[item] - 1 if item in customer_orders else inventory[item]
        for item in inventory
    }
    
    filtered_inventory = {
        item: qty for item, qty in updated_inventory.items() if qty > 0
    }
    return filtered_inventory

def calculate_order_statistics(customer_orders, products):
    total_ordered = len(customer_orders)
    percent_ordered = (total_ordered / len(products)) * 100
    return total_ordered, percent_ordered


def print_order_statistics(order_statistics):
    print("Order Statistics:", order_statistics)
    print("Total Products Ordered:", order_statistics[0])
    print("Percentage of Products Ordered:", f"{order_statistics[1]:.2f}%")

def print_updated_inventory(inventory):
    [print(f"{product}: {quantity}") for product, quantity in inventory.items()]


inventory = initialise_inventory(products)

customer_orders = get_customer_orders(inventory)

inventory = update_inventory(customer_orders, inventory)

order_stats = calculate_order_statistics(customer_orders, products)
print("\nOrder Statistics:")
print_order_statistics(order_stats)

print("\nUpdated Inventory:")
print_updated_inventory(inventory)

total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price:.2f}")
```
