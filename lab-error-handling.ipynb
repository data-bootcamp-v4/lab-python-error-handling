def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory

````python
def initialize_inventory(products):
    inventory = {}
    for product in products: 
        cantidad_valida = False
        while not cantidad_valida:
            try:
                cantidad = int(input(f"Ingresa la cantidad disponible de {product}: "))
                if cantidad < 0:
                    raise ValueError("Cantidad inválida. Ingresa un valor no negativo.")
                cantidad_valida = True
            except ValueError as error: 
                print(f"Error: {error}")
        inventory[product] = cantidad
    return inventory
````

````python
def calculate_total_price(customer_orders):
    prices = {}

    for product in customer_orders:
        while True:
            try:
                price = float(input(f"Introduce el precio del {product}: "))

                if price < 0: 
                    raise ValueError("El precio no puede ser negativo.")
                    continue
                prices.add(price)
                break

            except ValueError as error:
                print(f"Error: {error}. Por favor, introduce un valor válido.")

    total_price = sum(prices)
    return prices
````

````python
def get_customer_orders(inventory):
    customer_orders = set()
    while True:
        try:
            numero_pedido = int(input("¿Cuántos productos desea pedir? "))
            if numero_pedido <= 0:
                raise ValueError("El número de pedidos debe ser un número positivo.")
            break
        except ValueError:
            print(f"Por favor, introduce un número válido.")

    for n in range(numero_pedido):
        while True:
            
            nombre_producto = input(f"Introduce el nombre del producto {n + 1}: ")

            if nombre_producto not in inventory:
                    print("Ese producto no existe.")

            elif inventory[nombre_producto] <= 0:
                    print("No hay stock disponible.")
                
            customer_orders.add(nombre_producto)
            inventory[nombre_producto] -= 1
            break
    
    return customer_orders

````

````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = initialize_inventory(products)
print("Inicializando inventario")

customer_orders = get_customer_orders(inventory)
print("Pedidos del cliente.")

total_price = calculate_total_price(customer_orders)
print("Precio total.")
print("PEDIDO COMPLETADO")
print(f"Productos pedidos: {customer_orders}")
print(f"Precio total: {total_price:} €")
````
