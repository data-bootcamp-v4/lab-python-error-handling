## 1: Initialize inventory with error handling

```python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity >= 0:
                    inventory[product] = quantity
                    valid_input = True
                else:
                    print("Quantity cannot be negative. Please enter a valid quantity.")
            except ValueError:
                print("Invalid input. Please enter a valid quantity.")
    return inventory
```

## 2: Get customer orders with error handling

```python
def get_customer_orders(inventory):
    customer_orders = set()
    while True:
        try:
            product = input("Enter the name of a product that a customer wants to order: ").strip()
            if product not in inventory:
                raise ValueError("Product does not exist in the inventory.")
            if inventory[product] == 0:
                raise ValueError(f"{product} is out of stock.")
            customer_orders.add(product)
        except ValueError as error:
            print(f"Error: {error}")
            continue
        
        another = input("Do you want to add another product? (yes/no): ").strip().lower()
        if another != "yes":
            break
    return customer_orders
```

## 3። Calculate total price with error handling

```python
def calculate_total_price(customer_orders):
    total_price = 0
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                price = float(input(f"Enter the price of {product}: "))
                if price < 0:
                    raise ValueError("Price cannot be negative.")
                total_price += price
                valid_price = True
            except ValueError as error:
                print(f"Error: {error}")
    return total_price
```

## 4: Update inventory after orders

```python
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        inventory[product] -= 1
        if inventory[product] == 0:
            print(f"{product} is now out of stock.")
    return {product: qty for product, qty in inventory.items() if qty > 0}
```

## 5: Print order statistics and inventory

```python
def print_order_statistics(customer_orders, products):
    total_orders = len(customer_orders)
    percentage_unique = (total_orders / len(products)) * 100
    print("\nOrder Statistics:")
    print(f"Total Products Ordered: {total_orders}")
    print(f"Percentage of Unique Products Ordered: {percentage_unique:.1f}%")

def print_updated_inventory(inventory):
    print("\nUpdated Inventory:")
    for product, qty in inventory.items():
        print(f"{product}: {qty}")
```

## example 

```python
# Example list of products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# Step 1: Initialize inventory
inventory = initialize_inventory(products)

# Step 2: Get customer orders
customer_orders = get_customer_orders(inventory)

# Step 3: Calculate total price
total_price = calculate_total_price(customer_orders)

# Step 4: Update inventory
updated_inventory = update_inventory(customer_orders, inventory)

# Step 5: Print statistics and updated inventory
print_order_statistics(customer_orders, products)
print_updated_inventory(updated_inventory)

# Step 6: Print total price
print(f"\nTotal Price: {total_price:.2f}")
```
