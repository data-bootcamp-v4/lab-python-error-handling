```python
list_1 = [2,4,0,4,11,2602,36]
list_2 = [160,3,1719,19,13,-21]
```

```python
def find_outlier(para):
    odd_list = []
    even_list = []
    for x in para:
        if x % 2 == 0:
            even_list.append(x)
        else:
            odd_list.append(x)
    if len(odd_list) > len(even_list):
        return even_list[0]
    else:
        return odd_list[0]
```

```python
find_outlier(list_2)
```

```python
string_1 = "abc"
string_2 = "defer"
```

```python
def merge_strings(string_1, string_2):
    merged_str = ""
    len_1 = len(string_1)
    len_2 = len(string_2)
    smallest_str = min(len_1, len_2)
    if len_1 > smallest_str:
        remaining_str = string_1[smallest_str:]
    else:
        remaining_str = string_2[smallest_str:]
    a = string_1[:smallest_str]
    b = string_2[:smallest_str]
    for x,y in zip(a,b):
        merged_str += x + y
    merged_final = merged_str + remaining_str
    return merged_final
    
```

```python
merge_strings(string_1, string_2)
```

```python
string = "sTreSS"
```

```python
def first_non_repeating_letter(para):
    para_lower = para.lower()
    char_count = {}
    for x in para_lower:
        if x in char_count.keys():
            char_count[x] += 1
        else:
            char_count[x] = 1
    for let in para:
        if char_count[let.lower()] == 1:
            return let
```

```python
first_non_repeating_letter(string)
```
